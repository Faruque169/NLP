```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, it opened to the public on 15th May 1889, it ultimately vindicated itself aesthetically."
```

```python
import nltk
```

```python
words = nltk.word_tokenize(sentence)
```

```python
Tagged_elements = nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
nltk.ne_chunk(Tagged_elements).draw()
```
