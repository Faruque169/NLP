# **Wordnet Lemmatization**

### Lemmatization is a technique like stemming. The output we will get after lemmatization is called 'lema'. which is a root rather than a stem, the output stemming. After lemmatization, we will be getting a valid word that means the same thing.

###     NLTK provides wordNetLemmatizer class which is a thin wrapper around **WordNet CorpusReader** class to find lemma. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand the example-  

```python
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
nltk.download('wordnet')
```

```python
wordnet_lemmatizer = WordNetLemmatizer() # creating an instence
```

```python
''''
• n — Noun (wordnet.NOUN)
• v — Verb (wordnet.VERB)
• a — Adjective (wordnet.ADJ)
• r — Adverb (wordnet.ADV)
'''
wordnet_lemmatizer.lemmatize('going',pos='v')
```

```python
words = ['eat', 'eaten', 'ate','eats', 'writing','writes','history','programming','programs','finally','finalized']
```

```python
lemmatize_word = [word + ' ---> ' + (wordnet_lemmatizer.lemmatize(word, pos='v')) for word in words]
```

```python
lemmatize_word
```

```python
wordnet_lemmatizer.lemmatize('finally',pos='v')
```
